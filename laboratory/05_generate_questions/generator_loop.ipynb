```python
from mhdr.generator.generator_loop import (
    GenerationCSVStore,
    SeedCentroidScorer,
    FlanGenerator,
    GeneratorLoop
)

from pathlib import Path
import pandas as pd
```

```python
store = GenerationCSVStore(
    csv_path=Path("database.csv"),
    questions=Path("input/questions_master.csv")  
)

print("Database initialized.")
```

```python
df = store.load_df()
seed_df = df[df["status"] == "seed"]

scorer = SeedCentroidScorer(seed_df)

print("Centroid ready.")
```

```python
gen = FlanGenerator("google/flan-t5-base")
print("Generator loaded.")
```

```python
loop = GeneratorLoop(gen, store, scorer)
print("Loop ready.")
```

```python
loop.run(
    rounds=100,
    candidates_per_round=20,
    keep_ratio=0.3,
    seed_k=5,
)
```
