## 1. Overview

This lab consolidates survey questions from multiple mental health questionnaires into a single canonical dataset.

Each question is assigned a unique identifier (`qid`) based on its source questionnaire and item number, preserving the original question text for downstream analysis.

Survey questions are collected from **6 different sources**:
- CD-RISC (Connor-Davidson Resilience Scale)
- PERMA (Positive Emotions, Engagement, Relationships, Meaning, Accomplishment)
- PWB (Psychological Well-Being)
- PSS (Perceived Stress Scale)
- PWS (Psychological Well-Being Scale)
- UCLA (UCLA Loneliness Scale)

In total, there are **145 questions**.
The consolidated dataset is saved to `output/questions.csv` with the following columns:
- `qid`: Unique question identifier (e.g., CD_RISC_1)
- `text`: The question text
- `dataset`: Source questionnaire name




```python
from mhdr.dataloader.group_survey_question import load_questions_from_dir
from mhdr.dataloader.io import save_csv
from pathlib import Path

INPUT_DIR = Path.cwd() / "input"
OUTPUT_DIR = Path.cwd() / "output"
TEMP_DIR = Path.cwd() / "temp"

df = load_questions_from_dir(INPUT_DIR)
print(df.head())
print(df.shape)

save_csv(df, OUTPUT_DIR / "questions.csv") 
```
